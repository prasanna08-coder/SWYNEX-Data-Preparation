## 12. Date Preparation

The arrival date is represented across separate year, month,
and day columns.

A single datetime field is created to simplify chronological
analysis and downstream feature engineering.

The original date components are retained to preserve the
original information.

```python
month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["arrival_month_num"] = (
    df["arrival_date_month"].map(month_mapping)
)
```

```python
df["arrival_date"] = pd.to_datetime(
    {
        "year": df["arrival_date_year"],
        "month": df["arrival_month_num"],
        "day": df["arrival_date_day_of_month"]
    },
    errors="coerce"
)
```

## 14. Missing Value Treatment

Missing values are handled according to the meaning and role
of each variable.

The following assumptions are applied:

- `company`: Missing values indicate that no company information
  was associated with the booking. They are labeled as
  `Not Specified`.

- `agent`: Missing values indicate that an agent was not recorded.
  They are labeled as `Not Specified`.

- `country`: Missing values indicate that the country of origin
  was not available. They are labeled as `Unknown`.

- `children`: Only a very small number of records contain missing
  values. These values are replaced with 0 because the variable
  represents the number of children associated with the booking,
  and no child count was recorded for these observations.

These transformations preserve the records while making missing
information explicit.

```python
df["company"] = df["company"].fillna("Not Specified")
df["agent"] = df["agent"].fillna("Not Specified")
df["country"] = df["country"].fillna("Unknown")
df["children"] = df["children"].fillna(0)
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

## 15. Duplicate Removal

Exact duplicate records are removed from the working dataset.

Only rows that are identical across all available columns are
removed. Records that differ in at least one attribute are retained.

This approach avoids removing potentially valid bookings merely
because they share some characteristics with another booking.

```python
rows_before_duplicates = len(df)

df = df.drop_duplicates().copy()

rows_after_duplicates = len(df)

duplicates_removed = (
    rows_before_duplicates - rows_after_duplicates
)

print(f"Rows before duplicate removal: {rows_before_duplicates:,}")
print(f"Duplicate rows removed: {duplicates_removed:,}")
print(f"Rows after duplicate removal: {rows_after_duplicates:,}")
```

## 16. Handling Invalid ADR Values

The `adr` variable represents the average daily rate.

A negative ADR is not logically valid because a room rate cannot
be negative.

The dataset contains a negative ADR observation. Since this value
cannot represent a valid room rate, it is treated as missing
rather than assigning an arbitrary replacement value.

The value will therefore be imputed using the median ADR during
the preparation stage.

```python
df[df["adr"] < 0][
    ["hotel", "adr", "is_canceled", "reservation_status"]
]
```

```python
negative_adr_count = (df["adr"] < 0).sum()

print(
    f"Negative ADR records identified: {negative_adr_count}"
)
```

```python
df.loc[df["adr"] < 0, "adr"] = np.nan
```

```python
median_adr = df["adr"].median()

df["adr"] = df["adr"].fillna(median_adr)
```

```python
print("Negative ADR values remaining:", (df["adr"] < 0).sum())
print("Missing ADR values remaining:", df["adr"].isna().sum())
```
